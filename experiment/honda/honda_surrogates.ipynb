```python
import polars as pl
from sklearn.ensemble import GradientBoostingRegressor
import plotly.express as ex
```

```python
data = pl.read_parquet("./energy.parquet")
objs = data.columns[-11:]
objs
dec_vars = data.columns[:-11]
```

```python
data
```

```python
model = GradientBoostingRegressor().fit(data[dec_vars], data[objs[0]])
```

```python
fig = ex.scatter(
    x=model.predict(data[dec_vars]),
    y=data[objs[0]],
    labels={'x': 'Predicted values', 'y': 'Actual values'},
)

score = model.score(data[dec_vars], data[objs[0]])

# Add a diagonal line for reference
fig.add_scatter(
    x=[data[objs[0]].min(), data[objs[0]].max()],
    y=[data[objs[0]].min(), data[objs[0]].max()],
    mode='lines',
    line=dict(color='red', dash='dash'),
    showlegend=False
)

# Title
fig.update_layout(
    title=f'Predicted (gradient boosting) vs Actual values for {objs[0]} (R² = {score:.2f})',)
```
